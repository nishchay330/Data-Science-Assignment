```python
def analyze_numbers(numbers):
    even_count = 0
    odd_sum = 0

    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_sum += num

    return even_count, odd_sum


nums = [10, 15, 20, 25, 30, 35]

even_count, odd_sum = analyze_numbers(nums)

print("Count of even numbers:", even_count)
print("Sum of odd numbers:", odd_sum)
```

```python
def simple_interest(principal, rate=5, time=1):
    si = (principal * rate * time) / 100
    return si

# Calling with positional arguments
print("Positional Arguments:")
print("SI =", simple_interest(1000))          
print("SI =", simple_interest(1000, 8, 2))   

# Calling with keyword arguments
print("\nKeyword Arguments:")
print("SI =", simple_interest(principal=1000))
print("SI =", simple_interest(principal=1000, rate=10, time=3))
print("SI =", simple_interest(time=2, principal=1000, rate=7))
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        else:
            return "F"


student1 = Student("rajeev", 85)
student2 = Student("arsh", 72)

print("Student 1")
print("Name:", student1.name)
print("Marks:", student1.marks)
print("Grade:", student1.calculate_grade())

print("\nStudent 2")
print("Name:", student2.name)
print("Marks:", student2.marks)
print("Grade:", student2.calculate_grade())
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance!")

    def show_balance(self):
        print("Current Balance:", self.__balance)


account = BankAccount(1000)

account.show_balance()
account.deposit(500)
account.show_balance()
account.withdraw(300)
account.show_balance()
```

```python
try:
    file = open("numbers.txt", "r")

    numbers = []

    for line in file:
        numbers.append(float(line.strip()))

    file.close()

    total = sum(numbers)
    average = total / len(numbers)

    print("Total =", total)
    print("Average =", average)

except FileNotFoundError:
    print("Error: File does not exist.")
```
