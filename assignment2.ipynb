```python
name="educat#ion1234*$%"
vowels=0
consonents=0
num=0
special=0
for x in name:
    if x.isalpha():
        if x in("a","e","i","o","u"):
            vowels+=1
        else:
            consonents+=1
    elif x.isdigit():
        num+=1
    else:
        special+=1    
print("consonents:",consonents)
print("vowels:",vowels)
print("number:",num)
print("special:",special)
```
